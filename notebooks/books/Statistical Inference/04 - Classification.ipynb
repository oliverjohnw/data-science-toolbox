## ***4.3: Logistic Regression***

Logistic regression models the probability that y belongs to a particular category. We set a threshold to determine how we want to classify this prediction, usually choosing 0.5 as the threshold.

<br>

### ***4.3.1: The Logistic Model***

We must model $p(x)$ using a function that gives outputs between 0 and 1, for all values of x. To do so, we use the ***logistic function***:

$$
p(X) = \frac{e^{\beta_0 + \beta_1 X}}{1 + e^{\beta_0 + \beta_1 X}}
$$

To fit the model, we use a method called ***maximum likelihood***, which is discussed later. The figure below shows how our predictions are capped between 0 and 1:

![Logistic Curve](<img/04%20-%20Classification/logistic_curve.png>)

After a bit of manipulation of the formula above, we can calculate the ***odds***:

$$
\frac{p(X)}{1 - p(X)} = e^{\beta_0 + \beta_1 X}
$$

This odds value lies between 0 and infinity. Values of the odds close to 0 and infinity indicate very low and very high probabilities of the response, respectively. If, for example,  we receive a probability of 0.2, we would calculate the following odds:

$$
\text{Odds} = \frac{p(X)}{1 - p(X)} = \frac{0.2}{1 - 0.2} = \frac{0.2}{0.8} = 0.25
$$

And the interpretation of this is as follows: ***"For every 1 time the event happens, you'd expect it not to happen 4 times."***

A probability of 0.9 would have the odds:

$$
\text{Odds} = \frac{p(X)}{1 - p(X)} = \frac{0.9}{1 - 0.9} = \frac{0.9}{0.1} = 9
$$

And the interpretation of this is as follows: ***"For every 9 times the event happens, you'd expect it not to happen once."***

<br>

By taking the log of both sides, we arrive at the log odds, or logit:

$$
\log\left(\frac{p(X)}{1 - p(X)}\right) = \beta_0 + \beta_1 X
$$

The interpretation of this is as follows: ***Increasing X by one unit changes the log odds by $\beta_1$. AKA, we multiply the odds by $e^{\beta_1}$, when X increases by 1.*** Note that this relationship is not linear, and that's why it increases with multiplication and not addition. If $\beta_1$ is positive, an increase in X relates to an increase in $p(x)$. If $\beta_1$ is negative, an increase in X relates to an decrease in $p(x)$.

### ***4.3.2: Estimating Coefficients***

We use ***Maximum Likelihood*** to generate an estimate for the coefficients. This seeks etimates for the coefficients such that the predicted probability corresponds as closely as possible to the actual. Maximum likelihood is used to fit many non-linear methods, however, the scope of how it works is beyond this book.

The interpretation of significant predictors is similar to linear regression, except we use the ***z-stat*** rather than the t-stat.

Let's say we have a balance of $1000 and we want to predict the probability that we default. We would plug our coefficients in the formula: 

$$
p(X) = \frac{e^{\beta_0 + \beta_1 X}}{1 + e^{\beta_0 + \beta_1 X}}
$$

To get a probability of defaulting of 0.00576.

### ***4.3.4: Multiple Logistic Regression***

A ***confounding variable*** is an external variable that influences both the independent variable and the dependent variable, potentially giving the false impression that there's a causal relationship between the two.

For example, if we were to attempt to predict shark attacks based on ice cream sales we would see that theres a strong relationship between the two. However, we know that ice cream sales are largely correlated to the temperature. And when the temperature is higher, more people are in the water --- leading to more shark attacks. Without accounting for temperature, we might falsely assume that ice cream sales somehow cause shark attacks.

We must be concious of confounding variables, and remember that correlation != causation.
