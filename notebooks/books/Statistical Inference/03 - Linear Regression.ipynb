## ***3.1: Simple Linear Regression***

Simple linear regression involves modeling the relationship between a response Y with one predictor variable X. The equation is:

$$Y \approx \beta_0 + \beta_1 x_1 $$

Where:
* We assume a linear relationship between X and Y
* $\beta_0$ is the intercept
* $\beta_1$ is the slope

Both $\beta_0$ and $\beta_1$ are ***coefficients/parameters***, in which we use training data to tune and set. We use the training data such that we return parameters that estimate the linear regression line as close to the actual values as possible. There are many ways to measure "closeness", and the most common is through ***Least Squares***.

Least Squares chooses $\beta_0$ and $\beta_1$ to minimize the Residual Sum of Squares (RSS):

$$\hat{RSS} = e_1^2 + ... e_n^2$$

Where:
* $e_i^2$ is the residual values (actual - predicted) of the ith prediction.

Through a mathematical proof, we can show that the ***least squares coefficient estimates*** can be calculated with:

$$
\begin{align*}
\hat{\beta}_1 &= \frac{\sum_{i=1}^{n}(x_i - \bar{x})(y_i - \bar{y})}{\sum_{i=1}^{n}(x_i - \bar{x})^2} \\
\hat{\beta}_0 &= \bar{y} - \hat{\beta}_1 \bar{x}
\end{align*}
$$

***Interpretation***: Let's say we get the following linear regression estimate:

$$\hat{Y} = 7.03 + 0.0475 x_1 $$

The interpretation is as follows:
* For a 1 unit increase in X, we expect Y to increase by 0.0475 units. When $x_1$ is 0, we expect Y to be 7.03.

<br>

### ***3.1.2: Adressing the Accuracy of Coefficient Estimates***

If f is approximated to be linear, we can write the relationship as:

$$Y = \beta_0 + \beta_1 x_1 + \epsilon$$

Where:
* $\epsilon$ is a catch-all for what we miss with the model. This error may occur due to the fact that:
    - The true relationship may not be linear
    - Other non-included variables may cause the variation in Y
    - There may be measurement error
* We typically assume $\epsilon$ is independent of X

The equation above is the ***population regression line***. The least squares coefficient estimates characterize the ***least squares line***. In real life, the population regression line can never be calculated. But we can generate the least squares line to approximate this as best as possible.

In fact, this is fundamental to statistical problems. ***We are interested in calculating some population value. However, this is impossible. So we must generate our best estimate given our sample. The large the sample, the more confidence we have.***

The least squares coefficient estimates are considered to be un-biased. Think about how this relates to the concept of variance-bias above. We assume that these predictors do an adequate job of modeling the relationship. It must be noted that these can only be considered un-biased (adequate) if we are holding are assumptions.

In other words, what this means is that given a large enough sample (and holding our assumptions), we expect the estimates of $\beta_0 + \beta_1$ to be extremely close to the population regression line, if we were to generate these on many samples of data. On the first sample, we may overestimate these. On the second, we may underestimate. But given enough samples, it would average out to be very near the population coefficients.

<br>

#### ***Standard Error***

How can we measure how close we are to the population estimate? We can get an idea from the ***Standard Error***. The standard errors for $\beta_0$ and $\beta_1$ are:

$$
SE(\hat{\beta}_0)^2 = \sigma^2 \left[ \frac{1}{n} + \frac{\bar{x}^2}{\sum_{i=1}^{n}(x_i - \bar{x})^2} \right], \quad
SE(\hat{\beta}_1)^2 = \frac{\sigma^2}{\sum_{i=1}^{n}(x_i - \bar{x})^2}
$$

Remember, for all of these formulas to be valid, we need to assume that the errors $\epsilon$ have common variance. Often times this is not 100% true, but it can still be a good approximation.

In general, $\sigma^2$ is not known. So we can estimate $\sigma$ with the RSS:

$$
\sqrt{\frac{\text{RSS}}{n - 2}}
$$

<br>

#### ***Confidence Intervals***

Standard errors can be used to calculate ***Confidence Intervals***. A 95% confidence interval is defined as ***a range of values such that with 95% probability, the range will contain the true unknown value of the parameter***.

Confidence intervals have the following two properties:
* If we take repeated samples and construct CI's for each sample, 95% of the intervals will contain the true value for the parameter.
* For linear regression, the 95% CI for $\beta_1$ is:

$$\hat{\beta_1} \pm 2 * SE(\hat{\beta_1})$$


So if our $\hat{\beta_1}$ CI was [0.01, 0.25], we would say: ***"For a 1 unit increase in X1, we expect a 0.01 - 0.25 unit increase in Y".***

***Hypothesis Tests***

Standard errors can be used to perform ***hypothesis tests***. One hypothesis test may be:
* $H_o: \text{no relationship between x and y}$
* $H_a: \text{some relationship between x and y}$

Which in our simple linear regression example is:
* $H_o: \beta_1 = 0$
* $H_a: \beta_1 \neq 0$

To measure this, we can use a t-statistic, with 2-degress of freedom (in this example):

$$
t = \frac{\hat{\beta}_1 - 0}{SE(\hat{\beta}_1)}
$$

And from this, we can compute a ***p-value***. A small p-value means that it's unlikely to observe such a substantial association by random chance, and there full we reject the null. 

Notice in the calculation above that we divide by the standard error. The smaller the standard error, the large the t-value is, and the lower the p-value is. 

<br>


### ***3.1.3: Assessing Model Accuracy***

The quality of the fitted linear regression can be measured with:
* Residual  Standard Error (RSE)
* $R^2$
* Other metrics which we'll visit later.

Recall that associated with each value is an error term, $\epsilon$. Due to the presence, even if we know the true regression line, we would still never be able to perfectly predict Y from X.

The RSE is a measure off the standard deviation of $\epsilon$. It is, roughly speaking, the average amount the response will deviate from the true regression line. It can be computed with:

$$
\text{RSE} = \sqrt{\frac{1}{n - 2} \cdot \text{RSS}} = \sqrt{\frac{1}{n - 2} \sum_{i=1}^{n}(y_i - \hat{y}_i)^2}
$$

So if we were to obtain an RSE of 3.26, we could say: ***"Actual sales in each market deviate from the true regression line by approximately 3.26 units, on average"***. 

Whether or not this is acceptable depends on the context of the problem. RSE is considered to be a measure of the lack of fit of the model to the data.

The ***$R^2$*** is the proportion of variation in the response that is explained by the predictors. It falls between 0 and 1, since it is a proportion. The RSS can be measured with:

$$
R^2 = \frac{\text{TSS} - \text{RSS}}{\text{TSS}} = 1 - \frac{\text{RSS}}{\text{TSS}}
$$

Where:
* TSS is the Total Sum of Squares, defined by:

$$
\text{TSS} = \sum_{i=1}^{n} (y_i - \bar{y})^2
$$

The TSS measures the total variation in Y, AKA the amount of variability inherent in the response before regression. Hence, TSS - RSS measures the amount of variability in the response that is explained by performing regression, and dividing by the TSS makes it a proportion.

Values closer to 1 mean that a high proportion is explained. Values closer to 0 mean that a lower proportion is explained. This could be due to the fact that:
* The linear model is wrong
* $\sigma^2$ is high
* Both

The R2 can be challenging to interperet. If we know the relationship is expected to be linear, we expect the value to be close to 1. We must remember that R2 is a measure of the ***linear relationship between X and Y***. This is directly equivalent to the correlation, in a simple linear regression. However, when we add more features this does not hold. This is due to the fact that there is an association between features, and that affects the correlation.

## ***3.2: Multiple Linear Regression***

In multiple linear regression, the equation is represented by:

$$Y \approx \beta_0 + \beta_1 x_1 + ... +  \beta_p x_p$$

And the interpretation is (for $\beta_1$): "We expect an average increase of $\beta_1$ in Y for a 1 unit in crease in $x_1$, holding all other predictors constant".

<br>

### ***3.2.1: Estimating the Regression Coefficients***

Here, the RSS is defined as:

$$
\text{RSS} = \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 
= \sum_{i=1}^{n} \left( y_i - \hat{\beta}_0 - \hat{\beta}_1 x_{i1} - \hat{\beta}_2 x_{i2} - \cdots - \hat{\beta}_p x_{ip} \right)^2
$$

We are usually interested in answering the following questions:
* Is at least one of the predictors useful in predicting the response?
* Do all predictors help to explain Y, or is only a subset useful?
* How well does the model fit?
* Given a set of predictor values, what response values should we predict, and how accurate is this prediction?

<br>

#### ***Useful Predictors*** 

Just as before, we can use a hypothesis test to see if any of our predictors are related to the response:

* $H_o: \text{no relationship between the x's and y}$
* $H_a: \text{some relationship between the x's and y}$

Which in our simple linear regression example is:
* $H_o: \beta_1 = \beta_2 = ... \beta_p = 0$
* $H_a: \text{at least one} \beta_j \text{ is non-zero}$

For this, we can use the F statistic, which is calculated with:

$$
F = \frac{(\text{TSS} - \text{RSS}) / p}{\text{RSS} / (n - p - 1)}
$$

To accept the null, F must be close to 1. To reject, F must be greater than 1. We can directly compute an F-stat, but we can keep in mind that acception/rejection relies on values of n and p.
* For large values of n,  F does not need to be so large
* For small n, F needs to be large

<br>

#### ***Deciding Important Variables***

We can look at the individual p-vals provided, but if the number of features is high, this will likely lead to false discovery. We will often need to carry out ***feature selection*** here. The book offers 3 classical approaches:

1) ***Forward selection***: Start with no features, add variables that result in the lowest RSS. Stop when you hit a threshold.
2) ***Backward selection***: Start with all features, remove feature with the largest p-value.
3) ***Mixed selection***: Start with no variables, and add variables with the best fit. If at any point, a p-val resises above the threshold it will be removed.

Note that backward selection can not be used if p > n. Also, forward selection is greedy - meaning that it may include variables early that become redundant later. 

<br>

#### ***Assessing Model Fit***

It should be noted that the R2 will always increase as we add more variables. Adding in a feature that is not important or associated with the response tag will improve the R2, but will likely lead to overfitting.

<br>

#### ***Predictions***

There are three shorts of uncertainty with the predicitons.

Remember that $\hat{\beta}_0, \hat{\beta_1}, ... \hat{\beta_p}$ are estimates for $\beta_0, \beta_1, ... \beta_p$. Inaccuracy in these estimates is related to the reducible error.

We may also run into uncertainty due to the assumptions we make about the data. We assume that a linear model is appropriate for $f(x)$, when in reality, this is often times not true. This is what's known as model bias, and is potentially reducible error.

Finally, we must remember that $\epsilon$ will introduce random error in our prediction. We can use this error to generate ***Prediction Intervals***, which are just slightly wider than confidence intervals.

### ***3.3.2: Extensions of the Linear Model***

Standard linear regression gives interpretable results, and works quite well with some problems. However, it makes severely restrictive assumptions that are often times violated in practice. We are assuming that:
* The relationship between the predictors and the response are additive and linear. Additivity means that the association of the features between themselves and the response do not depend on the values of the other predictors.

The book offers a few classical approaches to relax these assumptions.

<br>

***Remove Additive Assumption***

In the equation:

$$Y = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + \epsilon$$

We assume a one unit increase in $x_1$ is associated with an average increase of $\beta_1$ in Y, holding $x_2$ constant. AKA the presence of $x_2$ does not affect this statement.

We can extend this by creating an ***interaction***:

$$Y = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + \beta_3 x_1 x_2 + \epsilon$$

And we can see now that the values of $x_1$ are effected by $x_2$. This removes the additivity assumption. Note that if we notice that the interaction between two variables is signficant, we should leave both features in the regression, even if they do not return significant. 

<br>

***Non-Linear Relationship***

To relax the assumption of a linear relationship, we can introduce polynomial regression:

$$Y = \beta_0 + \beta_1 x_1 + \beta_2 x_1^2 + \epsilon$$

![Non-Linear](<img/03%20-%20Linear%20Regression/nonlinear.png>)

Notice how, as we increase the degree of x in the plot above, we fit the model more specifically to the data. This is lowering the bias, while increasing the variance.

### ***3.3.3: Potential Problems***

The most common problems in fitting a linear regression are:
* Non-linearity of response-predictor relationships
* Correlation of error terms
* Non-constant variation of error terms
* Outliers
* High-leverage points
* Collinearity

#### ***Non-Linearity***

One way to address and examine if linearity exists in our data is to look at residual plots:

![residuals](<img/03%20-%20Linear%20Regression/residuals.png>)

When examining these, we hope to see no discernable pattern. In the figure above, we see that there is a strong pattern on the left and not much of a pattern on the right. This suggests that the data is not linear for the the left dataset.

A simple approach to fix this is by taking the log, square root, or square of the x variables.

#### ***Correlation of Error Terms***

One of the major assumptions we make with linear regression is assuming that the error values, $\epsilon$, are uncorrelated. If they are correlated, the standard error calculations we make will underestimate the true errors. In turn, this affects confidence intervals and p-vals, and inference. 

An error term may be correlated if we're working with time series data. One way to address and examine this is to plot the residuals as a function of time:

![residuals](<img/03%20-%20Linear%20Regression/residuals_time.png>)

Plotted above, we see that the first two plots seem to show no true pattern. We see that the residuals hover above and below 0, recommending that they're not correlated with each other. The values seem to change randomly from observation to observation.

In contrast, there is a pattern that emerges in the 3rd figure. We see elongated periods of negative residuals, followed be elongated periods of positives, and so on. This suggests that our error term is correlated. 

#### ***Non-Constant Variation of Error Terms***

This is also known as ***heteroskedacity***. 

![heteroskedacity](<img/03%20-%20Linear%20Regression/heteroskedacity.png>)

We can recognize heteroskedacity by the funnel shape on the left plot in the figure above.

A solution to this problem may be to transform Y. Notice how on the right, when we take the log(Y), we don't see that funnel shape. We may also take the square root of Y.

Another solution is to fit our model by ***weighted least squares***. 

#### ***Outliers***

Outlier points can affect the RSE and R2, and therefore our interpretation of the fit. These don't majorly impact the regression line, rather they affect the interpretation.

To examine this, we can plot the residuals.

![Outliers](<img/03%20-%20Linear%20Regression/outliers.png>)

In the figure above, we see that the point 20 is way off from the other residuals. This recommends that this point may be an outlier.

To quantify this, we can calculate ***studentized residuals***. This is calculating by dividing each residual by it's estimated standard error. Anything that's greater than 3 is most likely an outlier. This is shown in the rightmost plot of the figure above.

#### ***High-Leverage Point***

Whereas outliers are for Y values that are off, high-leverage points refer to X values that may be off. Whereas outliers don't really affect the regression line, high-leverage points do. 

These are a bit trickier to catch, as we often work in high dimension. The book recommends introducing a statistic called the ***leverage statistic***. 

#### ***Collinearity***

Collinearity occurs when two or more predictors are closely related. 

The figure below plots two regression fits. The fit on the left is for 2 variables that are not collinear, and the fit on the right is for 2 variables that are collinear.

![collinearity](<img/03%20-%20Linear%20Regression/collinearity.png>)

We can see that on the left, the estimates have more "room to vary". Meaning that a slight movement will not majorly affect our parameter estimates. However, on the right, a slight movement will cause a major affect on our parameter estimates. 

In more technical terms, collinearity reduces the accuracy of our coefficient estimates and causes standard errors to grow. This affects the t-stat that is calculated in the hypothesis test, and makes it very difficult to determine significant relationships.

One way to examine collinearity is to examine a correlation matrix. This works well with small datasets, but with large data this becomes impossible. 

When working with large datasets, we can calculate the ***Variance Inflation Factor***, which is a measure of the variance of $\hat{\beta_j}$ when fitting the entire model vs the variance when fitting alone. Values above 5 or 10 can be considered problematic.

There are two solutions when we determine a set of variables are collinear. We can:
* Drop one of the columns or
* Combine the features into one

## ***3.5: Comparison of Linear Regression to KNN***

Linear regression is a parametric approach, whereas KNN is non-parametric. 

KNN will calculate the K-nearest points, and take the average of the target values. The average is then our prediction. 

Just like the classification version from the previous chapter, a low k will fit the data very specifically (more flexible - low bias, high variance).

A parametric approach will outperform a non-parametric if the parametric form selected is close to the true form of $f$. So, for example, if the true relationship is linear, a linear regression will outperform KNN. 

It's important to note that KNN will perform worse than linear regression when in high dimensions.
