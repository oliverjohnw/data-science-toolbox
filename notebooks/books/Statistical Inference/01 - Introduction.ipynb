# ***Chapter 1: Introduction***

<br>

***Statistical learning*** refers to the vast set of tools for understanding data problems.

We can split these problems into two subsections:
* Supervised Learning
* Unsupervised Learning

***Supervised learning*** involves building a model used for predicting/forecasting an output, based on one or more inputs. In contrast, ***unsupervised learning*** does not contain any information on the output variable. Rather, it just uses the input variables. This is most often used for understanding relationships and patterns within the data.

<br>

***History of Statistical Learning***

A brief timeline of statistical learning throughout history:
- `1800's`: least squares + linear regression
- `1936`: Linear Discriminant Analysis (LDA)
- `1940's`: Logistic Regression
- `1970's`: Generalized Linear Model
- `1980's`: Classification + Regression Trees, Generalized Additive Models, Neural Networks
    - shifting away from linearity
- `1990's`: Support Vector Machines

<br>

***4 Premises of This Book***

This book relies on the following 4 premises:
1) Many statistical learning methods are relevant and useful in a wide range of problems, beyond just statistical sciences.
2) Statistical learning should not be considered as a series of black boxes. No approach will perform well in all applications, so you must understand the in's and outs.
3) It's important to know how the algorithm works and what it's made of, but it's not 100% necessary to know how to construct it.
4) This book assumes the reader wants to apply techniques to real world problems.
