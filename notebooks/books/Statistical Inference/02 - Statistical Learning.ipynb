## ***2.1: What is Statistical Learning?***

The relationship between y and X can be written:

$$Y = f(x) + \epsilon$$

Where:
* $f$ is some fixed, but unknown function of $X_1,..X_p$
* $\epsilon$ is a random error term, independent of X and mean 0
* $f$ represents the systematic information that $X$ provides about $y$

Note that in real life, $f$ is unknown. However, we build a model to represent this relationship as close as possible. This model is constructed based on historical data points. ***Statistical learning*** is the set of approaches for estimating $f$.

<br>

### ***2.1.1: Why Estimate f?***

There are two main reasons we may want to estimate $f$:
* Prediction
* Inference

<br>

#### ***Prediction***

We can predict Y by using:

$$\hat{y} = \hat{f}(x)$$

Sometimes, $\hat{f}(x)$ may be treated as a black box, in the sense that one is not typically concerned with the exact form, as long as it accurately predicts y.

The accuracy of $\hat{y}$ depends on two quantities:
* Reducible error
* Irreducible error

In reality, $\hat{f}(x)$ will almost never be a perfect representation of $f$. This error is ***reducible***, because we can potentially reduce the error by using a better method.

However, even if we formed a perfect estimate of $f$, our prediction would still have some error. This is because $Y$ is a function of $\epsilon$, which, by defintion, can not be predicted by $X$. So variability associated with $\epsilon$ effects our accuracy, and this is known as the ***irreducible error***. 

So why is the irreducible error larger than zero? Two reasons:
* We likely are missing factors/features that are useful for predicting $Y$.
* $\epsilon$ may also contain unmeasurable variation.

<br>

#### ***Inference***

We are sometimes interested in understanding the association between $Y$ and $X1,..X_n$. In this instance, we wish to estimate f, but not for the purpose of making predictions. Now, $\hat{f}(x)$ can ***absolutely not be treated like a black box***.

We may be interested in the following questions:
* Which predictors are associated with the response?
* What is the relationship between the response and the predictors?
* Can the relationship between Y and each predictor be summarized using a linear equation, or is it more complicated?

A general rule of thumb is that ***linear models are simple, which leads to interpretable inference and lower accuracy*** while ***non-linear methods are more complex, less interpretable, and have higher accuracy***.

<br>

### ***2.1.2: How Do We Estimate f?***

Our goal is to apply a statistical learning method to the training data, in order to estimate the function $f$. Most methods for this task can be categorized as either:
* Parametric
* Non-Parametric

<br>

#### ***Parametric***

A parameteric approach involves a two-step model approach.

First, we must make an assumption about the functional form of $f$. In the example of a linear regression, we can assume that $f$ is linear in $X$. This can be represented by:

$$f(x)=\beta_0 + \beta_1 x_1 + ... \beta_p x_p$$

Once we have assumed that $f$ is linear, the problem of estimating $f$ becomes greatly simplified. Instead of having to estimate an entire arbitrary p-dimensional function $f(x)$, we only have to estimate coefficients $\beta_0, \beta_1, ..., \beta_p$.

After the model has been selected, we need a procedure that uses the training data to fit/train the model. In this case, training the model involves estimating the coefficients $\beta_0, \beta_1, ..., \beta_p$. (The most common approach for linear regression is via OLS, however, there are many other approaches). 

The ***disadvantage*** to this approach is that it's likely the model we choose will not usually match the true, unknown form of $f$. If it's too far from $f$, then our estimates will be poor. 

We try to handle this by fitting a "flexible" method. In general, this means that the model can fit many different function forms of $f$, but requires a greater number of parameters. The more flexible the method, the more likely it will ***overfit***.

<br>

#### ***Non-Parametric***

These methods do not make explicit assumptions about the functional form of $f$. Rather, they seek to estimate $f$ so that it gets as close to the datapoints as possible (without overfitting). These methods have a higher potential of accurately fitting a wide range of possible shapes for $f$.

The ***disadvantage*** has been mentioned before, but to drill it home - these methods are prone to overfitting. Additionally, we do not reduce the problem of estimating f to a small number of parameters. Therefore, we ***need a large number of observation*** to train these methods. These methods are also sometimes harder to interperate.

<br>

## ***2.2: Assessing Model Accuracy***

When trying to solve a data problem, one should remember the ***No Free Lunch Theorem***: there is not one model that performs best on all possible datasets. We must decide whihc methods may be best, and attempt to fit these.

<br>


### ***2.2.1: Measuring the Quality of Fit***

The most commonly used metric for regression is the Mean Squared Error (MSE):

$$\text{MSE} = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$$

***The MSE will be small if predicted responses are close to the actual values.***

As you increase model flexibility, the ***degrees of freedom*** will increase. The figure below shows that, as the degrees of freedom for a model increases, the model will fit the training data better:

![Flexibility MSE](<img/02%20-%20Statistical%20Learning/flexibility_mse.png>)

However, this may lead to overfitting. The model may fit too specifically to the training data, and may not be able to generalize to the testing data very well. The model may be picking up on noise and random patterns within the training dataset, that are not representative of the population. We can suspect ***we are overfitting when we see a small train MSE, and a large test MSE*** (although there is always expected to be a slight difference).

The plot above also shows that, in general, as flexibility increases:
* ***The train MSE will decrease***.
* ***The test MSE will initially decrease, then increase***.

<br>

### ***2.2.2: The Bias-Variance Tradeoff***

The U-shape in the test MSE is the result of two competing methods:
* Bias
* Variance

The test MSE can be decomposed into the sum of these two, plus the error of the error terms $\epsilon$:

$$\text{Expected MSE} = \text{Variance} + \text{Bias}^2 + Var(\epsilon)$$

In order to minimize test error, we ***must select a method that achieves both***:
* ***low variance***
* ***low bias***

<br>

***Variance***

The variance is the amount by which $\hat{f}$ would change if we estimated it using a different training dataset. This comes with the knowledge that a new training set/different data will result in a newly fitted $\hat{f}$. Ideally, we do not the estimate of $\hat{f}$ to vary much from dataset to dataset.

The interpretation is as follows:
* High variance: small changes in the data lead to large changes in $\hat{f}$.
* More flexible methods have higher variance.

Refer to the figure above. We see that the flexible green curve is fit very specifically. An additional data point will cause a major change. Whereas if we had a linear regression, if we added a new datapoint it would result in a very minor change. 

<br>

***Bias***

The bias is the error introduced by approximating f by a much simpler model. For example, reference the figure below. On the left, we see a linear relationship displayed in the data, whereas on the right it is non-linear. 

If we were to fit a linear regression to these both, the left plot would have ***high bias*** and the right would have ***low bias***. 

![Linear vs Non-Linear](<img/02%20-%20Statistical%20Learning/linear_nonlinear.png>)

The idea is that no matter how much training data we have for the right, the data is simply non-linear. A linear regression will not do the trick.

Generally, more flexible options have lower bias. Also generally, as the model increases in flexibility, the variance will increase and the bias will decrease. However, the ***relative rate of change between these two*** will determine how the test MSE increases/decreases.

The figure below shows the bias and variance for 3 fitted models:

![Bias-Variance](<img/02%20-%20Statistical%20Learning/bias_variance.png>)

Notice how as flexibility increases:
* Bias decreases
* Variance increases

Also, the bias tends to initially decrease faster than the variance increased (specifically in the 1st and 3rd plot). This is related to a ***sharp decline in the MSE***. However, at some point increasing flexibility has little impact on the bias, but starts to significantly increase variance. This results in the ***test MSE increasing***.

This relationship is known as the ***bias-variance*** tradeoff. It's not actually possible to compute both bias and variance IRL, however, we need to keep these in mind as we develop our models. 

<br>

### ***2.2.3: The Classification Setting***

For classification settings, the most common approach for quantifying the accuracy of $\hat{f}$ is the ***training error rate***:

$$\text{Training Error Rate} = \frac{1}{n} \sum_{i=1}^{n} \mathbb{I}(\hat{y}_i \ne y_i)$$

Where:
* $\mathbb{I}(\hat{y}_i \ne y_i)$ is an indicator variable that equals 1 if $\hat{y} \ne y$

<br>

#### ***Bayes Classifier***

The test error rate is minimized by a simple classifier that assigns each observation to the most likely class, given it's predictor values. This is known as the Bayes Classifier. 

This rule says that we should assign a test observation with predictor vector $x_0$ to the class j for which:

$$\Pr(Y = j \mid X = x_0)$$

is the largest.

In a two class problem, we are presented with two probabilities:
* Probability of class 1
* Probability of class 2

We simply choose the probability which is the largest. 

From this method, we can generate a ***Bayes Decision Boundary***, which is shown below:

![Bayes Decision Boundary](<img/02%20-%20Statistical%20Learning/bayes_decision_boundary.png>)

In this figure:
* The orange region is where $\Pr(Y = orange \mid X)$ is greater than 50%
* The blue region is where $\Pr(Y = blue \mid X)$ is greater than 50%
* The purple line is where both are 50%, and is the Bayes Decision Boundary line

The Bayes Classifier produces the lowest possible test error rate, called the ***Bayes error rate***. Since it always chooses the class for whic the PR the largest, the error rate is:

$$1 - \mathbb{E} \left( \max_j \Pr(Y = j \mid X) \right)$$


This error rate is analagous to the irreducible error. The example above has an error rate of 0.133. This is because we see some overlap between the two group. If no overlap was shown, it would be 0.0.

Unfortunately, for real data we do not know the conditional probability distribution of Y given X, so computing the Bayes error rate is impossible. However, this serves as the unatainnable gold standard to compare other methods to.

<br>

#### ***K-Nearest Neighbors***

There are several approaches which attempt to estimate conditional probability of y given X, then classify based on this probability. KNN is one of these.

The KNN Classifier first identifies the K points in the training data that are closest to $X_{0}$ (represented as $N_{0}$). It then estimates the conditional probability for class j as the fraction of points in $N_{0}$ whose response values equal j:

$$\Pr(Y = j \mid X = x_0) = \frac{1}{K} \sum_{i \in \mathcal{N}_0} \mathbb{I}(y_i = j)$$

Then it classifies the test observation to the class with the largest probability. This is shown in the left figure below. In this example, there are two blue circles, adn one orange circle. The KNN Classifier would predict this observation as a blue circle.

![KNN](<img/02%20-%20Statistical%20Learning/KNN.png>)

Also shown above is the ***KNN decision boundary***.  This is the boundary in which K is set equal to 3, and calculated for all datapoints. 

The KNN classifier is often close to the optimal Bayes classifier. However, the choice of K has a big effect:
* Low k = flexible (low bias, high variance)
* High k = less flexible

This is shown below:

![KNN](<img/02%20-%20Statistical%20Learning/knn_k.png>)
