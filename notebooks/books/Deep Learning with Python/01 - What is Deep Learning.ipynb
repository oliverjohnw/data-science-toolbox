### ***1.1.1: Artificial Intelligence***

The term Artificial Intelligence (AI) came to be in the 1950's when we began to ask the question: "Can we make a computer think?". The technical definition of AI is: ***"the effort to automate intellectual tasks normally performed by humans"***. 

This includes:
* Machine Learning
* Deep Learning
* Hard-Coded Rules

<br>

### ***1.1.2: Machine Learning***

The traditional computational approach is to write a set of rules, and to get the results. Machine Learning (ML) works differently. With ML, we usually have the results, and we are interested in getting the rules.

In other words, we are looking for a way to meaningfully map our input data to a set of results. This is referred to as ***training***. We define a system, or a model, like Linear Regression or a Neural Network. We then ***train*** the model by setting the coefficients/parameters of that model. These are set by examining the training data, and returning the set that maximizes performance. 

ML and statistics are very similar, however, ML methods are used for large, complex datasets whereas statistical methods often work better on smaller datasets. There is also less mathematical theory to ML models, as it's more hands-on and empiral than statistics. Lastly, ML methods often times provide better prediction accuracy, wherears statistical methods often times provide better inference. 

<br>

### ***1.1.3: Learning Rules + Representations of Data***

So to summarize: ML transforms input data into meaningful outputs. This is "learned" from exposures to known examples of inputs and outputs. The central problem of ML is to***meaningfully transforming input data to outputs***.

In order for this to happen, ML requires:
* Input data
* Output data (labels)
* A measure of performance

Here's a very simple example:

![Example One](<img/01/example_one.png>)

In this example:
* The inputs are the coordinates of points
* The outputs are the color of the points
* The measure of performance will be a goodness-of-fit metric which checks how well our system is performing

When we solve this with ML, we are aiming for a system, or a new representation of our data, that cleanly seperates the white points from the black. In this problem, it's quite simple. We can draw a straight line to seperate the points, and our system can tell us that all observations to one side will be white, and all observations to the other side will be black.

In reality, it's never this simple. As data moves to higher-dimensions, finding that representation becomes increasingly difficult. 

So what if we systematically search for different sets of representations of the data? And use feedback from each set to indentify which representations look the best. This is called ***learning***!!!!

ML algorithms have a pre-defined set of transformations, and this is called a ***hypothesis space***. We want to try different trasnformations within this hypothesis space, and return the set that performs the best.

<br>

### ***1.1.4: The Deep in Deep Learning***

***Deep*** in deep learning refers to the successive layers of representation. Specifically, Deep Learning (DL) puts an emphasis on learning these successive layers with increasingly meaningful representations. The number of layers is referred to as the ***depth***.

Other appropriate names for DL could have been *"layered representations learning"* or *"hierarchical representations learning""*. 

Modern DL involves tens to hundreds of succesive layers. These are all learned automatically from exposure to training data. ML typically focuses on 1-2 layers, and is sometimes referred to as ***shallow learning***.
